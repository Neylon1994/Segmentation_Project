```python
#### Step 1: Adding Additonal Features to the Dataset

### We have no information on when this data was collected
### To decide age and length of time in the company we are going to use the max starting date of the customer

### Reference Date

Reference_Date = pd.to_datetime(df['Dt_Customer'],format='%d-%m-%Y').max()

#### Age of Customers
df['Age']= Reference_Date.year-df['Year_Birth']

##### Length of Tenure (How long they have been with the company)

df['Length_of_Time']=(Reference_Date - pd.to_datetime(df['Dt_Customer'],format='%d-%m-%Y')).dt.days

#### Combining Spend Together
df['Total_Product_Spend'] = df['MntWines'] + df['MntFruits'] + df['MntMeatProducts'] + df['MntFishProducts'] + df['MntSweetProducts'] + df['MntGoldProds']

### Adding the Number of kids togethe
df['Total_Kids'] = df['Kidhome'] + df['Teenhome']

#### Food Spend

df['Food_Spend']= df['MntFruits'] + df['MntMeatProducts'] + df['MntFishProducts'] + df['MntSweetProducts']
```
