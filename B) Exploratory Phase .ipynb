```python
#### 2) Look at Distributions 

### 2A) Histograms

#### This helps us understand the distribution of the results
### Are the results Left Skewed, Right Skewed or Evenly Distributed

features = df.columns.drop(['ID','Education','Year_Birth','Marital_Status','Dt_Customer'])

for fields in features: 

    plt.figure(figsize=(8,5))
    plt.hist(df[fields].dropna(), bins=20, edgecolor='black')
    plt.xlabel(fields)
    plt.ylabel('Frequency')
    plt.title(f'Histogram of {fields}')
    plt.show()
```

```python

### 2B) Boxplots

#### Helps us understand do some variables have many outliers


for boxplots in features:

    df.boxplot(boxplots)
    plt.title(f'Boxplot of {boxplots}')
    plt.ylabel('Value')
    plt.show()
```

```python
#### 3) Apply Correlation Matrix 

##### Helps us understand if there is redundanct variables being used
##### Have removed variables like ID,Education,Year_Birth,Marital_Status,Dt_Customer

correlation_heatmap = df[features].corr()

plt.figure(figsize=(16,8))
sns.heatmap(correlation_heatmap,annot=True,cmap='coolwarm',fmt='.2f')
```
